```python
import pandas as pd
from pathlib import Path
from urllib.request import urlretrieve

folder = Path('data/sms_spam_indonesia')
folder.mkdir(parents=True, exist_ok=True)
file = folder / 'dataset_sms_spam_v1.csv'

if not file.exists():
    url = 'https://raw.githubusercontent.com/Andikazidanef15/Sentiment-Analysis-on-Indonesian-SMS-Dataset/main/dataset_sms_spam_v1.csv'
    urlretrieve(url, file)

data = pd.read_csv(file).rename(
    columns={'Teks': 'teks', 'label': 'label_asli'}
)
data['label'] = data['label_asli'].map(
    {0: 'normal', 1: 'spam', 2: 'spam'}
)
data.head()
```

```python
import re

data['huruf_kecil'] = data['teks'].str.lower()
data['bersih'] = data['huruf_kecil'].str.replace(
    r'[^a-z0-9\s]', ' ', regex=True
)
data['bersih'] = data['bersih'].str.replace(
    r'\s+',' ', regex=True
).str.strip()

data[['teks', 'bersih']].head()
```

```python
import re

def case_folding(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Contoh Penggunaan
teks_awal = "Para mahasiswa sedang mempelajari teknik Preprocessing Data dalam pemrosesan bahasa alami. Mereka sangat antusias!"
hasil_case_folding = case_folding(teks_awal)

print("Hasil Case Folding:")
print(hasil_case_folding)
```

```python
import nltk
from nltk.tokenize import word_tokenize

# Unduh punkt dan punkt_tab jika belum pernah diunduh
nltk.download('punkt')
nltk.download('punkt_tab')

def tokenisasi(text):
    return word_tokenize(text)

# Contoh Penggunaan (menggunakan hasil dari Case Folding)
input_teks = "para mahasiswa sedang mempelajari teknik preprocessing data dalam pemrosesan bahasa alami mereka sangat antusias"
hasil_tokenisasi = tokenisasi(input_teks)

print("Hasil Tokenisasi:")
print(hasil_tokenisasi)
```

```python
!pip install Sastrawi
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory

def stopword_removal(tokens):
    factory = StopWordRemoverFactory()
    stopword = factory.create_stop_word_remover()

    # Memfilter list token
    filtered_tokens = [word for word in tokens if stopword.remove(word) != '']
    return filtered_tokens

# Contoh Penggunaan (menggunakan hasil dari Tokenisasi)
input_tokens = ['para', 'mahasiswa', 'sedang', 'mempelajari', 'teknik', 'preprocessing', 'data', 'dalam', 'pemrosesan', 'bahasa', 'alami', 'mereka', 'sangat', 'antusias']
hasil_stopword = stopword_removal(input_tokens)

print("Hasil Stopword Removal:")
print(hasil_stopword)
```

```python
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory

def stemming(tokens):
    factory = StemmerFactory()
    stemmer = factory.create_stemmer()

    # Mengubah setiap token ke kata dasar
    stemmed_tokens = [stemmer.stem(word) for word in tokens]
    return stemmed_tokens

# Contoh Penggunaan (menggunakan hasil dari Stopword Removal)
input_tokens = ['mahasiswa', 'mempelajari', 'teknik', 'preprocessing', 'data', 'pemrosesan', 'bahasa', 'alami', 'antusias']
hasil_stemming = stemming(input_tokens)

print("Hasil Stemming:")
print(hasil_stemming)
```
